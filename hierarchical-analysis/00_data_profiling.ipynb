##  Setup and Imports

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
from pathlib import Path
import os

# Configuration
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
plt.style.use('default')
sns.set_palette("husl")
warnings.filterwarnings('ignore')

# Data paths
DATA_PATH = Path('../pipeline-a-hierarchical/data')
print(f" Setup complete. Data path: {DATA_PATH}")
print(f" Data directory exists: {DATA_PATH.exists()}")
```

##  Dataset Inventory and File Analysis

```python
# Get all CSV files in the data directory
csv_files = list(DATA_PATH.glob('*.csv'))
csv_files.sort()

print(" PIPELINE A DATASET INVENTORY")
print("=" * 50)

file_info = []
for file_path in csv_files:
    try:
        # Get file size
        size_bytes = file_path.stat().st_size
        size_mb = size_bytes / (1024 * 1024)
        
        # Try to read first few rows to get basic info
        df_sample = pd.read_csv(file_path, nrows=1000)
        
        file_info.append({
            'filename': file_path.name,
            'size_mb': round(size_mb, 2),
            'sample_rows': len(df_sample),
            'columns': len(df_sample.columns),
            'status': ' Readable'
        })
        
        print(f" {file_path.name}")
        print(f"   Size: {size_mb:.2f} MB")
        print(f"   Columns: {len(df_sample.columns)}")
        print(f"   Sample rows loaded: {len(df_sample)}")
        print()
        
    except Exception as e:
        file_info.append({
            'filename': file_path.name,
            'size_mb': round(size_mb, 2) if 'size_mb' in locals() else 0,
            'sample_rows': 0,
            'columns': 0,
            'status': f' Error: {str(e)[:50]}...'
        })
        print(f" {file_path.name} - Error: {str(e)[:100]}...")
        print()

# Create summary DataFrame
file_summary = pd.DataFrame(file_info)
print("\n FILE SUMMARY TABLE")
print("=" * 30)
display(file_summary)
```

##  Quarterly Budget Data Analysis
### New Quarterly Spending Datasets

```python
print(" QUARTERLY BUDGET DATA ANALYSIS")
print("=" * 40)

try:
    # Load new quarterly budget datasets
    budget_func_quarterly = pd.read_csv(DATA_PATH / 'budget_functions_quarterly_all.csv')
    budget_subfunc_quarterly = pd.read_csv(DATA_PATH / 'budget_subfunctions_quarterly_all.csv')
    
    print(" BUDGET FUNCTIONS QUARTERLY DATA")
    print(f"   Records: {len(budget_func_quarterly):,}")
    print(f"   Columns: {len(budget_func_quarterly.columns)}")
    print(f"   Size: {budget_func_quarterly.memory_usage(deep=True).sum() / 1024**2:.2f} MB")
    print()
    
    print(" BUDGET SUBFUNCTIONS QUARTERLY DATA") 
    print(f"   Records: {len(budget_subfunc_quarterly):,}")
    print(f"   Columns: {len(budget_subfunc_quarterly.columns)}")
    print(f"   Size: {budget_subfunc_quarterly.memory_usage(deep=True).sum() / 1024**2:.2f} MB")
    print()
    
    # Time coverage analysis
    if 'fy' in budget_func_quarterly.columns:
        print(" QUARTERLY DATA TIME COVERAGE")
        print("-" * 30)
        func_years = budget_func_quarterly['fy'].agg(['min', 'max'])
        subfunc_years = budget_subfunc_quarterly['fy'].agg(['min', 'max'])
        
        print(f"Budget Functions: FY {func_years['min']} - {func_years['max']}")
        print(f"Budget Subfunctions: FY {subfunc_years['min']} - {subfunc_years['max']}")
        
        # Quarter distribution
        if 'quarter' in budget_func_quarterly.columns:
            print("\n QUARTER DISTRIBUTION:")
            quarter_dist = budget_func_quarterly['quarter'].value_counts().sort_index()
            for quarter, count in quarter_dist.items():
                print(f"   Q{quarter}: {count} records")
    
    # Spending amount analysis
    if 'amount' in budget_func_quarterly.columns:
        print("\n SPENDING AMOUNT ANALYSIS")
        print("-" * 25)
        
        # Remove null/zero amounts for analysis
        func_amounts = budget_func_quarterly['amount'].dropna()
        func_amounts = func_amounts[func_amounts > 0]
        
        if len(func_amounts) > 0:
            print(f"Total Spending Records: {len(func_amounts):,}")
            print(f"Total Amount: ${func_amounts.sum():,.0f}")
            print(f"Average per Record: ${func_amounts.mean():,.0f}")
            print(f"Median Amount: ${func_amounts.median():,.0f}")
        
        # Top spending functions by total
        print("\n TOP SPENDING FUNCTIONS (Quarterly Total):")
        top_functions = budget_func_quarterly.groupby('budget_function_name')['amount'].sum().sort_values(ascending=False).head(5)
        for func_name, total_amount in top_functions.items():
            print(f"   {func_name}: ${total_amount:,.0f}")
    
    # Display sample data
    print("\n BUDGET FUNCTIONS QUARTERLY SAMPLE")
    print("-" * 35)
    display(budget_func_quarterly.head())
    
    print("\n BUDGET SUBFUNCTIONS QUARTERLY SAMPLE")
    print("-" * 38)
    display(budget_subfunc_quarterly.head())
    
except Exception as e:
    print(f" Error loading quarterly budget data: {e}")
    print(" New datasets may not be available yet")
```

```python
print(" AGENCY DATA PROFILING")
print("=" * 30)

try:
    # Load agency data
    agencies = pd.read_csv(DATA_PATH / 'agency_ALL_FY.csv')
    
    print(f" Total Records: {len(agencies):,}")
    print(f" Columns: {len(agencies.columns)}")
    print(f" Memory Usage: {agencies.memory_usage(deep=True).sum() / 1024**2:.2f} MB")
    print()
    
    # Basic info
    print(" COLUMN ANALYSIS")
    print("-" * 20)
    print(agencies.info())
    print()
    
    # Display sample
    print(" SAMPLE DATA")
    print("-" * 15)
    display(agencies.head())
    
    # Time coverage
    if 'fy' in agencies.columns:
        print("\n TIME COVERAGE")
        print("-" * 15)
        fy_range = agencies['fy'].agg(['min', 'max'])
        print(f"Fiscal Years: {fy_range['min']} - {fy_range['max']}")
        
        fy_counts = agencies['fy'].value_counts().sort_index()
        print("\nRecords per Fiscal Year:")
        display(fy_counts)
    
    # Quarter analysis
    if 'fiscal_quarter' in agencies.columns:
        print("\n QUARTER DISTRIBUTION")
        print("-" * 20)
        quarter_counts = agencies['fiscal_quarter'].value_counts().sort_index()
        display(quarter_counts)
    
except Exception as e:
    print(f" Error loading agency data: {e}")
```

##  Data Quality Assessment

```python
print(" DATA QUALITY ANALYSIS")
print("=" * 30)

def analyze_data_quality(df, dataset_name):
    """Comprehensive data quality analysis"""
    print(f"\n {dataset_name.upper()}")
    print("-" * len(dataset_name))
    
    # Basic stats
    print(f"Shape: {df.shape}")
    print(f"Memory: {df.memory_usage(deep=True).sum() / 1024**2:.2f} MB")
    
    # Missing values
    missing = df.isnull().sum()
    missing_pct = (missing / len(df) * 100).round(2)
    
    if missing.sum() > 0:
        print("\n Missing Values:")
        missing_df = pd.DataFrame({
            'Count': missing[missing > 0],
            'Percentage': missing_pct[missing_pct > 0]
        })
        display(missing_df)
    else:
        print(" No missing values")
    
    # Duplicates
    duplicates = df.duplicated().sum()
    if duplicates > 0:
        print(f"\n  Duplicate rows: {duplicates} ({duplicates/len(df)*100:.2f}%)")
    else:
        print(" No duplicate rows")
    
    # Data types
    print("\n Data Types:")
    dtype_counts = df.dtypes.value_counts()
    for dtype, count in dtype_counts.items():
        print(f"   {dtype}: {count} columns")

# Analyze available datasets
datasets_to_analyze = [
    ('budget_functions.csv', 'Budget Functions'),
    ('budget_subfunctions.csv', 'Budget Subfunctions'),
    ('agency_ALL_FY.csv', 'Agencies'),
    ('federal_accounts_ALL_FY.csv', 'Federal Accounts')
]

for filename, name in datasets_to_analyze:
    file_path = DATA_PATH / filename
    if file_path.exists():
        try:
            df = pd.read_csv(file_path)
            analyze_data_quality(df, name)
        except Exception as e:
            print(f"\n Error analyzing {name}: {e}")
```

##  Relationship Analysis

```python
print(" DATASET RELATIONSHIPS")
print("=" * 25)

# Analyze relationships between datasets
try:
    # Load key datasets for relationship analysis
    budget_functions = pd.read_csv(DATA_PATH / 'budget_functions.csv')
    budget_subfunctions = pd.read_csv(DATA_PATH / 'budget_subfunctions.csv')
    agencies = pd.read_csv(DATA_PATH / 'agency_ALL_FY.csv')
    
    print(" KEY FIELD ANALYSIS")
    print("-" * 20)
    
    # Budget function codes
    if 'budget_function_code' in budget_functions.columns:
        bf_codes = set(budget_functions['budget_function_code'].astype(str))
        print(f"Budget Function codes: {len(bf_codes)}")
        print(f"Sample codes: {list(bf_codes)[:5]}")
    
    # Budget subfunction codes  
    if 'budget_subfunction_code' in budget_subfunctions.columns:
        bsf_codes = set(budget_subfunctions['budget_subfunction_code'].astype(str))
        print(f"Budget Subfunction codes: {len(bsf_codes)}")
        print(f"Sample codes: {list(bsf_codes)[:5]}")
    
    # Agency analysis
    if 'id' in agencies.columns:
        agency_ids = agencies['id'].nunique()
        print(f"Unique agencies: {agency_ids}")
    
    if 'code' in agencies.columns:
        agency_codes = agencies['code'].nunique()
        print(f"Unique agency codes: {agency_codes}")
    
    # Time coverage comparison
    print("\n TIME COVERAGE COMPARISON")
    print("-" * 25)
    
    if 'fy' in agencies.columns:
        agency_years = agencies['fy'].agg(['min', 'max'])
        print(f"Agency data: FY {agency_years['min']} - {agency_years['max']}")
    
except Exception as e:
    print(f" Error in relationship analysis: {e}")
```
